```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from utility import get_data
from sklearn.model_selection import train_test_split

# Load data
df = get_data()
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['StudentID', 'Semester', 'MDM', 'Stat', 'Formalise', 'Cost', 'Final_Exam_Score']), df['Final_Exam_Score'], test_size=0.3, random_state=42)
# convert everything to float
X_train = X_train.astype(float)
X_test = X_test.astype(float)
y_train = y_train.astype(float)
y_test = y_test.astype(float)
```

```python
# ==========================================
# 1. Linear Regression Model
# ==========================================
X_train_filled = X_train.fillna(0)
X_test_filled = X_test.fillna(0)
lr_model = LinearRegression()
lr_model.fit(X_train_filled, y_train)

# Predict and evaluate
lr_preds = lr_model.predict(X_test_filled)
print("--- Linear Regression ---")
print(f"MSE: {mean_squared_error(y_test, lr_preds):.4f}")
print(f"R2 Score: {r2_score(y_test, lr_preds):.4f}\n")
```

```python
# ==========================================
# 2. Random Forest Model
# ==========================================
rf_model = RandomForestRegressor(n_estimators=50, random_state=42)
rf_model.fit(X_train, y_train)

# Predict and evaluate
rf_preds = rf_model.predict(X_test)
print("--- Random Forest ---")
print(f"MSE: {mean_squared_error(y_test, rf_preds):.4f}")
print(f"R2 Score: {r2_score(y_test, rf_preds):.4f}")
```

```python
# ==========================================
# 3. XGBoost Model
# ==========================================
from xgboost import XGBRegressor

xgb_model = XGBRegressor(n_estimators=50, random_state=42)
xgb_model.fit(X_train, y_train)

# Predict and evaluate
xgb_preds = xgb_model.predict(X_test)
print("--- Random Forest ---")
print(f"MSE: {mean_squared_error(y_test, xgb_preds):.4f}")
print(f"R2 Score: {r2_score(y_test, xgb_preds):.4f}")
```
